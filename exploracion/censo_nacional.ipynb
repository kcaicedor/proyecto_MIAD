# Instalación de librerías y Extracción de data

```python
from pathlib import Path
import pandas as pd
```

```python
# Desde exploracion/censo_nacional.ipynb
base = Path("../data/censo_nacional")

tablas = []

for carpeta in sorted(base.iterdir()):
    if carpeta.is_dir():
        archivos = list(carpeta.rglob("*VIV*.csv"))

        if archivos:
            df = pd.read_csv(archivos[0])
            df["carpeta_origen"] = carpeta.name
            tablas.append(df)

censo_viviendas = pd.concat(tablas, ignore_index=True)

print("Filas cargadas:", len(censo_viviendas))
display(censo_viviendas.head())
```
